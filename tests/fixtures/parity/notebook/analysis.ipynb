# Analysis Notebook

This notebook demonstrates parity testing for ipynb cell attribution.

```python
def load_data() -> list:
    with open("data/input.txt") as f:
        return f.readlines()
```

```python
def process(rows: list) -> dict:
    counts = {}
    for row in rows:
        key = row.strip()
        counts[key] = counts.get(key, 0) + 1
    return counts
```

```python
data = load_data()
result = process(data)
print(result)
```
